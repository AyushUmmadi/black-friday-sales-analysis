# Black Friday Sales Data Analysis
## Part 5: Combining Gender & Marital Status

Further investigate the combined effect of Gender and Marital Status on purchasing behavior across age, product category, city category, and years stayed in the current city.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('BlackFriday.csv')

del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

Create a combined `Gender_MaritalStatus` column so we can analyze the two factors together in a single plot.

```python
df['Gender_MaritalStatus'] = df['Gender'] + "_" + df['Marital_Status'].astype(str)
df.head()
```

```python
sns.countplot(x=df['Gender_MaritalStatus'])
plt.title('Transactions by Gender + Marital Status')
plt.show()
```

```python
sns.countplot(x=df['Age'], hue=df['Gender_MaritalStatus'])
plt.title('Age Group, split by Gender + Marital Status')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x=df['Product_Category_1'], hue=df['Gender_MaritalStatus'])
plt.title('Product Category, split by Gender + Marital Status')
plt.show()
```

```python
sns.countplot(x=df['Stay_In_Current_City_Years'], hue=df['Gender_MaritalStatus'])
plt.title('Years Stayed in City, split by Gender + Marital Status')
plt.show()
```

```python
sns.countplot(x=df['City_Category'], hue=df['Gender_MaritalStatus'])
plt.title('City Category, split by Gender + Marital Status')
plt.show()
```

**Key takeaway:** Unmarried males (`M_0`) are consistently the largest segment across almost every cut of the data — by age group, product category, city category, and years in current city — reinforcing that this segment drives the bulk of Black Friday activity.
