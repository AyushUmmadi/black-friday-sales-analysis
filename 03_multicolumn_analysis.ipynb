# Black Friday Sales Data Analysis
## Part 3: Multi-Column Analysis

Combine Age, Gender, Marital Status, and City Category to analyze more complex relationships and derive insights from combinations of factors.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('BlackFriday.csv')

del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

### Age & Gender

```python
sns.countplot(x="Age", hue='Gender', data=df)
plt.title('Transactions by Age Group, split by Gender')
plt.show()
```

```python
sns.countplot(x="Gender", hue='Age', data=df)
plt.title('Transactions by Gender, split by Age Group')
plt.show()
```

### Gender & Marital Status

```python
sns.countplot(x="Gender", hue='Marital_Status', data=df)
plt.title('Transactions by Gender, split by Marital Status')
plt.show()
```

### City Category

```python
sns.countplot(x=df['City_Category'])
plt.title('Transactions by City Category')
plt.show()
```

```python
df.groupby('City_Category').size().plot(kind='pie', autopct='%0.1f')
plt.ylabel('')
plt.title('Share of Transactions by City Category')
plt.show()
```

```python
sns.countplot(x="City_Category", hue='Gender', data=df)
plt.title('City Category, split by Gender')
plt.show()
```

```python
df.groupby('City_Category').sum(numeric_only=True)['Purchase'].plot(kind='pie', autopct='%.1f')
plt.ylabel('')
plt.title('Share of Total Purchase Amount by City Category')
plt.show()
```

**Key takeaway:** Combining these factors shows purchasing behavior varies meaningfully by age and gender together, not just individually — and City Category B accounts for the largest share of both transactions and total spend.
