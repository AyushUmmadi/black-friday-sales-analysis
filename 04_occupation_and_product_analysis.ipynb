# Black Friday Sales Data Analysis
## Part 4: Occupation & Product Analysis

Analyze how occupation influences purchasing behavior and which product categories are most popular among different occupations. Also explores how long customers have stayed in their current city.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('BlackFriday.csv')

del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

### Stay in Current City (Years)

```python
sns.countplot(x=df['Stay_In_Current_City_Years'])
plt.title('Transactions by Years Stayed in Current City')
plt.show()
```

```python
sns.countplot(x=df['Stay_In_Current_City_Years'], hue='Gender', data=df)
plt.title('Years Stayed in Current City, split by Gender')
plt.show()
```

```python
sns.countplot(x=df['Stay_In_Current_City_Years'], hue='City_Category', data=df)
plt.title('Years Stayed in Current City, split by City Category')
plt.show()
```

```python
sns.countplot(x='City_Category', hue='Age', data=df)
plt.title('City Category, split by Age Group')
plt.show()
```

```python
df.groupby('Stay_In_Current_City_Years').size().plot(kind='pie', autopct='%.2f')
plt.ylabel('')
plt.title('Share of Transactions by Years Stayed in City')
plt.show()
```

```python
df.groupby('Stay_In_Current_City_Years').sum(numeric_only=True)['Purchase'].plot(kind='bar')
plt.title('Total Purchase Amount by Years Stayed in City')
plt.ylabel('Total Purchase')
plt.show()
```

### Occupation

```python
sns.countplot(x=df['Occupation'])
plt.title('Transactions by Occupation')
plt.show()
```

```python
df.groupby('Occupation').size().plot(kind='bar')
plt.title('Transaction Count by Occupation')
plt.show()
```

```python
# Total purchase amount by occupation, sorted lowest to highest
df.groupby('Occupation').sum(numeric_only=True)['Purchase'].sort_values().plot(kind='bar')
plt.title('Total Purchase Amount by Occupation')
plt.ylabel('Total Purchase')
plt.show()
```

```python
# Average purchase amount by occupation, sorted lowest to highest
df.groupby('Occupation').mean(numeric_only=True)['Purchase'].sort_values().plot(kind='bar')
plt.title('Average Purchase Amount by Occupation')
plt.ylabel('Average Purchase')
plt.show()
```

```python
sns.countplot(x='Occupation', hue='Marital_Status', data=df)
plt.title('Occupation, split by Marital Status')
plt.show()
```

```python
sns.countplot(x='Occupation', hue='Gender', data=df)
plt.title('Occupation, split by Gender')
plt.show()
```

```python
# Number of distinct products purchased by customers in each occupation
df.groupby('Occupation').nunique()['Product_ID'].sort_values().plot(kind='bar')
plt.title('Distinct Products Purchased, by Occupation')
plt.ylabel('Distinct Product Count')
plt.show()
```

**Key takeaway:** Purchasing volume and total spend vary noticeably across occupation codes, and some occupations show a much wider variety of products purchased than others — suggesting occupation is a useful signal for targeting.
